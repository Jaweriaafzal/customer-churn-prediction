```python
df = pd.read_csv(r"C:\Users\DELL\Downloads\ecommerce_customer_churn_dataset.csv")
df = df.drop_duplicates()  # keep cleaned data consistent with earlier step
```

```python
df.columns.tolist()
```

```python
# Feature Engineering

df["Purchase_Frequency"] = (
    df["Total_Purchases"] / (df["Membership_Years"] + 1)
)

df["Engagement_Score"] = (
    df["Login_Frequency"]
    + df["Pages_Per_Session"]
    + df["Social_Media_Engagement_Score"]
)

df["Customer_Activity_Score"] = (
    df["Login_Frequency"]
    + df["Email_Open_Rate"]
    + df["Mobile_App_Usage"]
)

df["Risk_Score"] = (
    df["Cart_Abandonment_Rate"]
    + df["Returns_Rate"]
    + df["Days_Since_Last_Purchase"]
)
```

```python
new_features = [
    "Purchase_Frequency",
    "Engagement_Score",
    "Customer_Activity_Score",
    "Risk_Score"
]

df[new_features].describe()
```

```python
df[new_features].head()
```

```python
correlation = df.corr(numeric_only=True)

churn_correlations = (
    correlation["Churned"]
    .drop("Churned")
    .sort_values()
)

churn_correlations
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 8))

churn_correlations.plot(kind="barh")

plt.title("Feature Correlation with Churn")
plt.xlabel("Correlation with Churn")
plt.ylabel("Features")
plt.show()
```

```python
from sklearn.model_selection import train_test_split

X = df.drop("Churned", axis=1)
y = df["Churned"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
numerical_features = X.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = X.select_dtypes(
    include=["object"]
).columns.tolist()

print("Numerical Features:")
print(numerical_features)

print("\nCategorical Features:")
print(categorical_features)
```

```python
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer

preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            StandardScaler(),
            numerical_features
        ),
        (
            "cat",
            OneHotEncoder(
                handle_unknown="ignore"
            ),
            categorical_features
        )
    ]
)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer

preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            Pipeline(steps=[
                ("imputer", SimpleImputer(strategy="mean")),
                ("scaler", StandardScaler())
            ]),
            numerical_features
        ),
        (
            "cat",
            Pipeline(steps=[
                ("imputer", SimpleImputer(strategy="most_frequent")),
                ("encoder", OneHotEncoder(handle_unknown="ignore"))
            ]),
            categorical_features
        )
    ]
)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

logistic_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", LogisticRegression(max_iter=1000))
    ]
)

logistic_model.fit(X_train, y_train)
```

```python
y_pred_logistic = logistic_model.predict(X_test)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

print("Accuracy:", accuracy_score(y_test, y_pred_logistic))
print("Precision:", precision_score(y_test, y_pred_logistic))
print("Recall:", recall_score(y_test, y_pred_logistic))
print("F1 Score:", f1_score(y_test, y_pred_logistic))
```

```python
print(classification_report(y_test, y_pred_logistic))
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

cm = confusion_matrix(y_test, y_pred_logistic)

plt.figure(figsize=(6, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues"
)

plt.title("Logistic Regression - Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

```python
# ================================
# RIDGE & LASSO REGRESSION
# Target: Lifetime_Value
# ================================
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import Ridge, Lasso
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np
import pandas as pd

target_reg = "Lifetime_Value"

# Churned ko bhi drop kar rahe hain (classification label, regression leakage se bachne ke liye)
X_reg = df.drop([target_reg, "Churned"], axis=1)
y_reg = df[target_reg]

X_train_r, X_test_r, y_train_r, y_test_r = train_test_split(
    X_reg, y_reg, test_size=0.20, random_state=42
)

numerical_features_r = X_reg.select_dtypes(include=["int64", "float64"]).columns.tolist()
categorical_features_r = X_reg.select_dtypes(include=["object"]).columns.tolist()

print("Numerical Features:", numerical_features_r)
print("Categorical Features:", categorical_features_r)

preprocessor_r = ColumnTransformer(
    transformers=[
        ("num", Pipeline(steps=[
            ("imputer", SimpleImputer(strategy="mean")),
            ("scaler", StandardScaler())
        ]), numerical_features_r),
        ("cat", Pipeline(steps=[
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("encoder", OneHotEncoder(handle_unknown="ignore"))
        ]), categorical_features_r)
    ]
)

# ---------- Ridge Regression ----------
ridge_model = Pipeline(steps=[
    ("preprocessor", preprocessor_r),
    ("model", Ridge(alpha=1.0))
])
ridge_model.fit(X_train_r, y_train_r)
ridge_preds = ridge_model.predict(X_test_r)

print("\nRidge Regression:")
print("  MSE :", mean_squared_error(y_test_r, ridge_preds))
print("  RMSE:", np.sqrt(mean_squared_error(y_test_r, ridge_preds)))
print("  R2  :", r2_score(y_test_r, ridge_preds))

# ---------- Lasso Regression ----------
lasso_model = Pipeline(steps=[
    ("preprocessor", preprocessor_r),
    ("model", Lasso(alpha=0.1))
])
lasso_model.fit(X_train_r, y_train_r)
lasso_preds = lasso_model.predict(X_test_r)

print("\nLasso Regression:")
print("  MSE :", mean_squared_error(y_test_r, lasso_preds))
print("  RMSE:", np.sqrt(mean_squared_error(y_test_r, lasso_preds)))
print("  R2  :", r2_score(y_test_r, lasso_preds))
```

```python
# ================================
# DECISION TREE CLASSIFIER
# Target: Churned
# ================================
from sklearn.tree import DecisionTreeClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    accuracy_score, classification_report, confusion_matrix,
    ConfusionMatrixDisplay
)
import matplotlib.pyplot as plt

dt_model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("model", DecisionTreeClassifier(max_depth=5, random_state=42))
])

dt_model.fit(X_train, y_train)
dt_preds = dt_model.predict(X_test)

print("Decision Tree Accuracy:", accuracy_score(y_test, dt_preds))
print("\nClassification Report:\n", classification_report(y_test, dt_preds))

cm = confusion_matrix(y_test, dt_preds)
disp = ConfusionMatrixDisplay(confusion_matrix=cm)
disp.plot(cmap="Blues")
plt.title("Decision Tree - Confusion Matrix")
plt.show()
```

```python
# ================================
# DECISION TREE VISUALIZATION
# ================================
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt

# Feature names nikaalna (preprocessing ke baad wale names)
feature_names = (
    numerical_features +
    list(dt_model.named_steps["preprocessor"]
         .named_transformers_["cat"]
         .get_feature_names_out(categorical_features))
)

plt.figure(figsize=(20, 10))
plot_tree(
    dt_model.named_steps["model"],
    feature_names=feature_names,
    class_names=["Not Churned", "Churned"],
    filled=True,
    rounded=True,
    fontsize=8,
    max_depth=3   # zyada deep tree confusing hoti hai, top 3 levels dikha rahe hain
)
plt.title("Decision Tree - Churn Prediction")
plt.show()
```

```python
# ================================
# STEP 8: Compare Logistic Regression vs Decision Tree
# ================================
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score
)
import pandas as pd

# Probabilities nikaalna (ROC-AUC ke liye zaroori hai)
logistic_probs = logistic_model.predict_proba(X_test)[:, 1]
dt_probs = dt_model.predict_proba(X_test)[:, 1]

comparison_table = pd.DataFrame({
    "Model": ["Logistic Regression", "Decision Tree"],
    "Accuracy": [
        accuracy_score(y_test, logistic_preds),
        accuracy_score(y_test, dt_preds)
    ],
    "Precision": [
        precision_score(y_test, logistic_preds),
        precision_score(y_test, dt_preds)
    ],
    "Recall": [
        recall_score(y_test, logistic_preds),
        recall_score(y_test, dt_preds)
    ],
    "F1": [
        f1_score(y_test, logistic_preds),
        f1_score(y_test, dt_preds)
    ],
    "ROC-AUC": [
        roc_auc_score(y_test, logistic_probs),
        roc_auc_score(y_test, dt_probs)
    ]
})

print(comparison_table.round(3))
```

```python
# ================================
# STEP 9: Overfitting Visualization
# (Train vs Validation Accuracy across tree depths)
# ================================
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt

depths = range(1, 21)
train_scores = []
val_scores = []

for depth in depths:
    model = Pipeline(steps=[
        ("preprocessor", preprocessor),
        ("model", DecisionTreeClassifier(max_depth=depth, random_state=42))
    ])
    model.fit(X_train, y_train)

    train_preds = model.predict(X_train)
    val_preds = model.predict(X_test)

    train_scores.append(accuracy_score(y_train, train_preds))
    val_scores.append(accuracy_score(y_test, val_preds))

# Plot
plt.figure(figsize=(10, 6))
plt.plot(depths, train_scores, marker="o", label="Training Accuracy")
plt.plot(depths, val_scores, marker="o", label="Validation Accuracy")
plt.xlabel("Tree Depth (max_depth)")
plt.ylabel("Accuracy")
plt.title("Overfitting Analysis: Training vs Validation Accuracy")
plt.xticks(list(depths))
plt.legend()
plt.grid(alpha=0.3)
plt.show()

# Overfitting point find karna (jahan gap sabse zyada badhne lagta hai)
gap = [t - v for t, v in zip(train_scores, val_scores)]
best_depth = val_scores.index(max(val_scores)) + 1
print(f"Best validation accuracy at depth = {best_depth} (val acc = {max(val_scores):.3f})")
print(f"\nTrain-Val gap at each depth:")
for d, g in zip(depths, gap):
    print(f"  depth={d}: gap={g:.3f}")
```

```python
# ================================
# STEP 10: Feature Importance
# Decision Tree vs Logistic Regression
# ================================
import pandas as pd
import matplotlib.pyplot as plt

feature_names = (
    numerical_features +
    list(dt_model.named_steps["preprocessor"]
         .named_transformers_["cat"]
         .get_feature_names_out(categorical_features))
)

# ---------- Decision Tree Feature Importance ----------
dt_importance = pd.DataFrame({
    "Feature": feature_names,
    "Importance": dt_model.named_steps["model"].feature_importances_
}).sort_values(by="Importance", ascending=False).head(10)

# ---------- Logistic Regression Coefficients ----------
logistic_importance = pd.DataFrame({
    "Feature": feature_names,
    "Coefficient": logistic_model.named_steps["model"].coef_[0]
})
logistic_importance["AbsCoefficient"] = logistic_importance["Coefficient"].abs()
logistic_importance = logistic_importance.sort_values(by="AbsCoefficient", ascending=False).head(10)

# ---------- Side-by-Side Plot ----------
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

axes[0].barh(dt_importance["Feature"], dt_importance["Importance"], color="#4C72B0")
axes[0].invert_yaxis()
axes[0].set_title("Decision Tree - Feature Importance")
axes[0].set_xlabel("Importance")

colors = ["#DD8452" if c > 0 else "#C44E52" for c in logistic_importance["Coefficient"]]
axes[1].barh(logistic_importance["Feature"], logistic_importance["Coefficient"], color=colors)
axes[1].invert_yaxis()
axes[1].set_title("Logistic Regression - Coefficients")
axes[1].set_xlabel("Coefficient (sign shows direction)")

plt.tight_layout()
plt.show()

# ---------- Overlap Check ----------
top_dt_features = set(dt_importance["Feature"].head(5))
top_log_features = set(logistic_importance["Feature"].head(5))
common = top_dt_features.intersection(top_log_features)

print("Top 5 Decision Tree features:", top_dt_features)
print("Top 5 Logistic Regression features:", top_log_features)
print(f"\nCommon features in both top-5 lists: {common}")
print(f"Overlap: {len(common)}/5")
```
